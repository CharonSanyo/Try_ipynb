```python
course = "Python"
scores = [72, 85, 91, 68, 88, 95]
student_count = len(scores)

print("变量已经创建")
```

```python
average = sum(scores) / student_count

print("课程：", course)
print("人数：", student_count)
print("平均分：", average)
print(f"平均分：{average:.2f}")
```

```python
bonus = 5
adjusted_scores = [score + bonus for score in scores]
adjusted_scores
```

```python
import pandas as pd

df = pd.DataFrame({
    "student": ["Alice", "Bob", "Carol", "David", "Eva", "Frank"],
    "score": scores
})

df
```

```python
df["passed"] = df["score"] >= 80
df
```

```python
df[df["passed"]]
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))
plt.bar(df["student"], df["score"], color="steelblue")
plt.axhline(80, color="orange", linestyle="--", label="Target: 80")

plt.title("Student Scores")
plt.xlabel("Student")
plt.ylabel("Score")
plt.ylim(0, 100)
plt.legend()
plt.show()
```

```python
threshold = 90

selected = df[df["score"] >= threshold]

print(f"达到 {threshold} 分的人数：{len(selected)}")
display(selected)

display(df.head(3))
display(df.describe())
```
